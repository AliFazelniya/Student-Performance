```python
import pandas as pd
import scripts.plot_utils
import scripts.plot_utils as plot_utils
```

```python
df1 = pd.read_csv("../data/raw/student-mat.csv" , encoding="utf-8", sep=";")
df2 = pd.read_csv("../data/raw/student-por.csv" , encoding="utf-8", sep=";")
```

```python
df1.head()
```

```python
df1.info()
```

```python
df1.isnull().sum()
```

```python
numeric_df = df1.select_dtypes(include=["int64", "float64"])

Q1 = numeric_df.quantile(0.25)
Q3 = numeric_df.quantile(0.75)
IQR = Q3 - Q1

outliers = (numeric_df < (Q1 - 1.5 * IQR)) | (numeric_df > (Q3 + 1.5 * IQR))
outliers.sum().sort_values(ascending=False)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(15, 6))
sns.boxplot(data=numeric_df)
plt.xticks(rotation=45)
plt.show()
```
